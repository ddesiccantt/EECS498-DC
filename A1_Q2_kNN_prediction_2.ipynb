```python
import torch

def knn_predict_one(train_points, train_labels, query, k=3):
    distances = ((train_points - query) ** 2).sum(dim=1).sqrt()

    _, indices = distances.topk(k, largest=False)
    neighbor_labels = train_labels[indices]

    prediction = torch.bincount(neighbor_labels).argmax()
    return prediction

train_points = torch.tensor([
    [0., 0.],
    [1., 0.],
    [0., 1.],
    [5., 5.],
    [6., 5.]
])
train_labels = torch.tensor([0, 0, 0, 1, 1])

# result = knn_predict_one(
#     train_points,
#     train_labels,
#     torch.tensor([5., 4.])
# )

# print(result)
queries = torch.tensor([
    [1., 1.],
    [5., 4.],
    [6., 6.]
])

results = []

for query in queries:
    prediction = knn_predict_one(train_points, train_labels, query, k=3)
    results.append(prediction)

predictions = torch.stack(results)

print(results)
print(predictions)

true_labels = torch.tensor([0, 1, 1])

accuracy = (predictions == true_labels).float().mean()

print(accuracy)
```
