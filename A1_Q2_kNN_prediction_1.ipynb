```python
# kNN_prediction_1_practice
import torch

# AI_generated_points
train_points = torch.tensor([
    [0., 0.],
    [1., 0.],
    [0., 1.],
    [5., 5.],
    [6., 5.]
])

train_labels = torch.tensor([0, 0, 0, 1, 1])

query = torch.tensor([1., 1.])
k = 3

distances = ((train_points - query) ** 2).sum(dim=1).sqrt()

#print(distances)

_, indices = distances.topk(k, largest=False)
#print(indices)
neighbor_labels = train_labels[indices]

#print("neighbor_labels", neighbor_labels)

predictions = torch.bincount(neighbor_labels).argmax()

print("predictions:", predictions)
```
