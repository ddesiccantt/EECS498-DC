```python
import torch

train = torch.tensor([
    [0., 0.],
    [3., 4.],
    [6., 8.]
])  # [3, 2]：3 个训练点

queries = torch.tensor([
    [0., 0.],
    [3., 0.]
])  # [2, 2]：2 个查询点

q = queries.unsqueeze(1)
t = train.unsqueeze(0)

diff = q - t
print(diff)

squared_distances = (diff ** 2).sum(dim=2)
print(squared_distances)

_, indices = squared_distances.topk(k=2, dim=1, largest=False)
print(indices)

train_labels = torch.tensor([0, 1, 1])

neighbor_labels = train_labels[indices]

print(neighbor_labels)
# tensor([[0, 1],
#         [0, 1]])

results = []

for labels in neighbor_labels:
    counts = torch.bincount(labels)
    prediction = counts.argmax()
    results.append(prediction)

predictions = torch.stack(results)

print(predictions)
# tensor([0, 0])
```
