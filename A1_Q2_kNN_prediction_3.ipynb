```python
import torch

train_points = torch.tensor([
    [0., 0.],
    [1., 0.],
    [0., 1.],
    [5., 5.],
    [6., 5.]
])

train_labels = torch.tensor([0, 0, 0, 1, 1])

def knn_predict_one(train_points, train_labels, query, k=3):
    distances = ((train_points - query) ** 2).sum(dim=1).sqrt()

    _, indices = distances.topk(k, largest=False)
    neighbor_labels = train_labels[indices]
    prediction = torch.bincount(neighbor_labels).argmax()

    return prediction

def knn_predict(train_points, train_labels, queries, k=3):
    results = []

    for query in queries:
        prediction = knn_predict_one(
            train_points, train_labels, query, k
        )
        results.append(prediction)
        # print("results",results)

    return torch.stack(results)

queries = torch.tensor([
    [1., 1.],
    [5., 4.],
    [6., 6.],
    [0., 0.]
])

predictions = knn_predict(
    train_points, train_labels, queries, k=3
)

# print(predictions)
# tensor([0, 1, 1])

val_points = torch.tensor([
    [0.5, 0.5],
    [5., 4.],
    [6., 6.]
])

val_labels = torch.tensor([0, 1, 1])

best_k = None
best_accuracy = -1

for k in [1, 3, 5]:
    predictions = knn_predict(
        train_points, train_labels, val_points, k=k)
    # print("predictions",predictions)

    accuracy = (predictions == val_labels).float().mean()
    # print(accuracy)
    # print(f"k={k}, accuracy={accuracy.item():.2%}")
    if accuracy >= best_accuracy:
        best_accuracy = accuracy
        best_k = k

# print("best_k:", best_k)
# print("best_accuracy:", best_accuracy)

# 新的测试样本，没有用于选择 k
test_points = torch.tensor([
    [0.2, 0.3],
    [5.5, 5.],
    [6., 4.]
])
test_labels = torch.tensor([0, 1, 1])

test_predictions = knn_predict(
    train_points, train_labels, test_points, k=best_k
)

test_accuracy = (
    (test_predictions == test_labels).float().mean().item()
)

print(test_predictions)
print(f"测试准确率：{test_accuracy:.2%}")
```
