```python
import torch

# 第一张：所有像素值为 0，表示纯黑
image1 = torch.zeros(3, 32, 32)
print(image1)
# 第二张：所有像素值为 1，表示纯白
# 这里约定像素值范围是 0～1
image2 = torch.ones(3, 32, 32)

# 新增第 0 轴，表示“图片编号”
batch = torch.stack([image1, image2], dim=0)

print(batch)
# torch.Size([2, 3, 32, 32])
```

```python
points = torch.tensor([
    [1., 2.],
    [4., 6.],
    [1., 5.]
])  # shape: [3, 2]

query = torch.tensor([1., 2.])  # 待比较的点，shape: [2]
diff = points - query
print(diff)
sdis = (diff**2).sum(dim=1)
print(sdis)
d = sdis.sqrt()
print(d)
```

```python
distances = torch.tensor([3.2, 1.1, 2.4, 0.8])

values, indices = distances.topk(k=3, largest=False)

print(values)   # tensor([0.8000, 1.1000])
print(indices)  # tensor([3, 1])
labels = torch.tensor([7, 9, 4, 9])

neighbor_labels = labels[indices]

print(neighbor_labels)
# tensor([9, 9])

neighbor_labels = torch.tensor([9, 9, 4])

counts = torch.bincount(neighbor_labels)

print(counts)
# tensor([0, 0, 0, 0, 1, 0, 0, 0, 0, 2])
prediction = counts.argmax()
print(prediction)
```
