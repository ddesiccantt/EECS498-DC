```python
import torch

def knn_predict_one(train_points, train_labels, query, k=3):
    distances = ((train_points - query) ** 2).sum(dim=1).sqrt()

    _, indices = distances.topk(k, largest=False)
    neighbor_labels = train_labels[indices]
    prediction = torch.bincount(neighbor_labels).argmax()

    return prediction

def knn_predict(train_points, train_labels, queries, k=3):
    results = []

    for query in queries:
        prediction = knn_predict_one(
            train_points, train_labels, query, k
        )
        results.append(prediction)

    return torch.stack(results)

train_images = torch.stack([
    torch.zeros(3, 2, 2),        # 黑色
    torch.ones(3, 2, 2),         # 白色
    torch.ones(3, 2, 2) * 0.5    # 灰色
])

image_labels = torch.tensor([0, 1, 2])

# [3, 3, 2, 2] → [3, 12]
train_flat = train_images.flatten(start_dim=1)
# print(train_flat)

# query_image = torch.ones(3, 2, 2) * 0.4
# query_flat = query_image.flatten()  # [12]

# prediction = knn_predict_one(
#     train_flat, image_labels, query_flat, k=1
# )

# print(prediction)
# # tensor(1)

query_images = torch.stack([
    torch.ones(3, 2, 2) * 0.1,
    torch.ones(3, 2, 2) * 0.9,
    torch.ones(3, 2, 2) * 0.4
])

# 每张图片分别展开
queries_flat = query_images.flatten(start_dim=1)

print(queries_flat.shape)
# torch.Size([3, 12])

predictions = knn_predict(
    train_flat, image_labels, queries_flat, k=1
)

print(predictions)
# tensor([0, 1, 2])
```
